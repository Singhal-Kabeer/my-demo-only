```python
from bpetokenizer import *
tokenizer = load_bpe_tokenizer()
```

```python
import json

def tokenize_and_save(dataset, output_filename):
    tokenized = []
    for doc in dataset:
        d = doc.copy()
        d["tokens"] = tokenizer.encode(d["text"])
        d.pop("text")
        tokenized.append(d)
        
    with open(output_filename, "w", encoding="utf-8") as f:
        json.dump(tokenized, f)

tokenize_and_save(train_data, "local_tokenized_train.json")
tokenize_and_save(dev_data, "local_tokenized_dev.json")
tokenize_and_save(test_data, "local_tokenized_test.json")
```
