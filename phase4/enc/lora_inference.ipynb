```python
class Phase4ClassificationModel(nn.Module):
    def __init__(self, base_model, num_classes=2, pad_token_id=0):
        super().__init__()
        self.base_model = base_model
        self.num_classes = num_classes
        self.pad_token_id = pad_token_id
        
        self.classification_head = nn.Linear(self.base_model.cfg.d_model, num_classes)

    def forward(self, idx):
        encoder_out = self.base_model.encode(idx)
        
        mask = (idx != self.pad_token_id).float().unsqueeze(-1)
        sum_embeddings = (encoder_out * mask).sum(dim=1)
        real_token_counts = mask.sum(dim=1)
        
        pooled_output = sum_embeddings / torch.clamp(real_token_counts, min=1e-9)
        
        logits = self.classification_head(pooled_output)
        
        return logits

class LoRALinear(nn.Module):
    def __init__(self, base_layer: nn.Linear, r: int = 8, alpha: float = 8.0):
        super().__init__()
        self.base = base_layer
        self.scale = alpha / r
        
        for param in self.base.parameters():
            param.requires_grad = False
            
        in_dim = base_layer.in_features 
        out_dim = base_layer.out_features
        
        self.lora_A = nn.Linear(in_dim, r, bias=False)
        self.lora_B = nn.Linear(r, out_dim, bias=False)
        
        nn.init.zeros_(self.lora_B.weight)

    def forward(self, x):
        return self.base(x) + self.scale * self.lora_B(self.lora_A(x))
```

```python
device = torch.device("mps" if torch.backends.mps.is_available() else "cuda" if torch.cuda.is_available() else "cpu")
cfg = EncDecConfig(vocab_size=50260)
base_lm = EncoderDecoderLM(cfg)

model = Phase4ClassificationModel(base_lm, num_classes=2, pad_token_id=0)

rank = 8
alpha_val = 8.0
for block in model.base_model.encoder_blocks:
    block.attn.q_proj = LoRALinear(block.attn.q_proj, r=rank, alpha=alpha_val)
    block.attn.v_proj = LoRALinear(block.attn.v_proj, r=rank, alpha=alpha_val)

model.load_state_dict(torch.load("checkpoint.pt", map_location=device))
model.to(device)
model.eval()
```

```python
_, _, test_data = load_cjpe()
```

```python
predictions = []

with torch.no_grad():
    for doc in test_data:
        doc_id = doc.get('id')
        encoded_doc = doc['tokens'][-1024:]
        input_tensor = torch.tensor(encoded_doc, dtype=torch.long).unsqueeze(0).to(device)
        logits = model(input_tensor)
        pred_val = torch.argmax(logits, dim=1).item()
        predictions.append({"id": doc_id, "pred": pred_val})

with open("prediction.json", "w", encoding="utf-8") as f:
    json.dump(predictions, f, indent=4)
```
