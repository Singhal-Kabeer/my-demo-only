```python
TRAIN_PATH = "Phase_7_PCR/Train/train.json"
TRAIN_CANDIDATES_PATH = "Phase_7_PCR/Train/train_candidates.json"
DEV_PATH = "Phase_7_PCR/Dev/dev.json"
DEV_CANDIDATES_PATH = "Phase_7_PCR/Dev/dev_candidates.json"
TEST_PATH = "Phase_7_PCR/Test/test.json"
TEST_CANDIDATES_PATH = "Phase_7_PCR/Test/test_candidates.json"
```

```python
HF_TRAIN_PATH = hf_hub_download(repo_id=HF_DATASET_REPO, filename=TRAIN_PATH, repo_type="dataset", token=HF_TOKEN)
HF_TRAIN_CANDIDATES_PATH = hf_hub_download(repo_id=HF_DATASET_REPO, filename=TRAIN_CANDIDATES_PATH, repo_type="dataset", token=HF_TOKEN)
HF_DEV_PATH = hf_hub_download(repo_id=HF_DATASET_REPO, filename=DEV_PATH, repo_type="dataset", token=HF_TOKEN)
HF_DEV_CANDIDATES_PATH = hf_hub_download(repo_id=HF_DATASET_REPO, filename=DEV_CANDIDATES_PATH, repo_type="dataset", token=HF_TOKEN)
HF_TEST_PATH = hf_hub_download(repo_id=HF_DATASET_REPO, filename=TEST_PATH, repo_type="dataset", token=HF_TOKEN)
HF_TEST_CANDIDATES_PATH = hf_hub_download(repo_id=HF_DATASET_REPO, filename=TEST_CANDIDATES_PATH, repo_type="dataset", token=HF_TOKEN)
```

```python
with open(HF_TEST_PATH,"r", encoding="utf-8") as f:
    test_data = json.load(f)
with open(HF_TEST_CANDIDATES_PATH,"r", encoding="utf-8") as f:
    test_candidates_data = json.load(f)

with open(HF_TRAIN_PATH,"r", encoding="utf-8") as f:
    train_data = json.load(f)
with open(HF_TRAIN_CANDIDATES_PATH,"r", encoding="utf-8") as f:
    train_candidates_data = json.load(f)

with open(HF_DEV_PATH,"r", encoding="utf-8") as f:
    dev_data = json.load(f)
with open(HF_DEV_CANDIDATES_PATH,"r", encoding="utf-8") as f:
    dev_candidates_data = json.load(f)
```

```python
train_queries = dict()
train_labels = dict()
train_candidates = dict()

for key in train_data:
    text = " ".join(train_data[key]['text'])
    train_queries[key] = np.array(tokenizer.encode(text), dtype=np.uint16)
    train_labels[key] = train_data[key]['relevant_candidates']

for key in train_candidates_data:
    text = " ".join(train_candidates_data[key]['text'])
    train_candidates[key] = np.array(tokenizer.encode(text), dtype=np.uint16)

output = {"queries": train_queries, "cands": train_candidates, "labels": train_labels}

with open(f"tok_train.pkl", "wb") as f:
    pickle.dump(output, f)
```
